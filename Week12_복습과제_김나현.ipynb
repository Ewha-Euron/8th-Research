- **파이프라인(pipeline) 함수**
    - 문장 분류, 문장 생성, 토큰 분류 등 다양한 작업에 대한 전처리, 모델 아키텍처, 후처리 각각 처리 가능.
- 파이프라인 클래스
    - 입력된 작업(task)에 모델(model)로 적합한 파이프라인 구축
    - 작업 매개변수 -> 수행하려는 작업 의미
    - `text_generation` & `gpt2` 입력
    - `TextGenerationPipeline` 클래스 인스턴스 생성
- 입력 텍스트(`text_inputs`) : 생성하려는 문장의 입력 문맥 전달.
- 최대 길이`(max_length`) : 생성될 문장의 최대 토큰 수 제한
- 반환 시퀀스 개수(`num_return_sequences`) : 생성할 텍스트 시퀀스의 수.
- 패딩 토큰 ID(`pad_token_id`) : 자유 생성(Open-end generation) 여부 설정함. 해당 매개변수 사용하면 모델이 입력된 문장의 문맥 기반으로 자유롭게 다음 단어나 문장 생성함.

```python
# CoLA 데이터셋 불러오기

import torch
from torchtext.datasets import CoLA
from transformers import AutoTokenizer
from torch.utils.data import DataLoader

def collator(batch, tokenizer, device):
    source, labels, texts = zip(*batch)
    tokenized = tokenizer(
        texts,
        padding="longest",
        truncation=True,
        return_tensors="pt"
    )
    inputs_ids = tokenized["input_ids"].to(device)
    attention_mask = tokenized["attention_mask"].to(device)
    labels = torch.tensor(labels, dtype=torch.long).to(device)
    return inputs_ids, attention_mask, labels

train_data = list(CoLA(split="train"))
valid_data = list(CoLA(split="dev"))
test_data = list(CoLA(split="test"))

tokenizer = AutoTokenizer.from_pretrained("gpt2")
tokenizer.pad_token = tokenizer.eos_token

epoch = 3
batch_size = 16
device = "cuda" if torch.cuda.is_available() else "cpu"

train_loader = DataLoader(
    train_data,
    batch_size=batch_size,
    collate_fn=lambda x: collator(x, tokenizer, device),
    shuffle=True,
    )

valid_loader = DataLoader(
    valid_data,
    batch_size=batch_size,
    collate_fn=lambda x: collator(x, tokenizer, device),
    )

test_loader = DataLoader(
    test_data,
    batch_size=batch_size,
    collate_fn=lambda x: collator(x, tokenizer, device),
    )
```

- **CoLA 데이터셋**
    - train, dev, test로 구성
    - train 데이터셋 : 모델 학습
    - dev 데이터셋 : 모델 검증
    - test 데이터셋 : 학습된 모델의 성능 평가에 사용됨.

- GPT-2는 사전 학습 시 패딩 기법 사용하지 않기 때문에 토크나이저의 특수 토큰 중 패딩 토큰이 따로 포함되어 있지 않음.
    - 문장 분류 모델 학습하기 위해 문장의 끝 의미하는 eos 토큰으로 패딩 토큰 대체.

- `collator` 함수 : 배치를 토크나이저로 토큰화, 패딩(padding), 절사(truncation), 반환 형식 설정(return_tensors) 작업 수행함.
    - 패딩에 인자를 `longest`로 설정하면 가장 긴 시퀀스에 대해 패딩 적용
    - 절사에 인자를 `True`로 설정하면 입력 시퀀스 길이가 최대 길이 초과하는 경우 해당 시퀀스 자름.
    - 반환 형식 설정에 `pt` 입력하면 파이토치 텐서 형태로 결과 반환함.

- 토크나이저 -> 토큰 ID(input_ids)와 어텐션 마스크(attention_mask) 반환.
    - GPT 모델은 어텐션 마스크 이용하여 입력 문장에서 패딩 부분 무시하고 실제 단어에 대해 처리함.
- 토큰ID는 토크나이저가 각 토큰에 대해 부여한 숫자 ID 담고 있음.
- 어텐션 마스크는 입력 문장의 실제 단어에 대응하는 부분은 1, 패딩에 대응하는 부분은 0으로 채움.

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
```

```python
# GPT-2 모델 설정
from torch import optim
from transformers import GPT2ForSequenceClassification

model = GPT2ForSequenceClassification.from_pretrained(
    pretrained_model_name_or_path="gpt2",
    num_labels=2
).to(device)

model.config.pad_token_id = model.config.eos_token_id
optimizer = optim.Adam (model.parameters(), lr=5e-5)
```

- GPT-2 문장 분류 모델(GPT2ForSequenceClassification)
    - GPT-2 기반으로 하는 시퀀스 분류 모델
    - 기본 GPT-2 모델과 유사하게 작동
    - 최종 출력 계층이 분류를 위해 미세 조정됨.
    - 사전 학습 시 토큰의 패딩 기법 사용하지 않으므로 토크나이저에 패딩 토큰 포함 x
    - 문장 분류 모델 학습 위해서는 모델이 고정된 길이의 입력 필요로 하므로 문장 끝을 나타내는 eos 토큰 사용하여 패딩 토큰으로 대체.

```python
# GPT-2 모델 학습 및 검증
import numpy as np
from torch import nn

def calc_accuracy(preds, labels):
    pred_flat = np.argmax(preds, axis=1).flatten()
    labels_flat = labels.flatten()
    return np.sum(pred_flat == labels_flat) / len(labels_flat)

def train(model, optimizer, dataloader):
    model.train()
    train_loss = 0.0
    for inputs_ids, attention_mask, labels in dataloader:
        outputs = model(
            input_ids=inputs_ids,
            attention_mask=attention_mask,
            labels=labels
        )
        loss = outputs.loss
        train_loass += loss.item()

        optimizer.zero()
        loss.backward()
        optimizer.step()

    train_loss =  train_loss / len(dataloader)
    return train_loss

def evaluation(model, dataloader):
    with torch.no_grad():
        model.eval()
        criterion = nn.CrossEntropyLoss()
        val_loss, val_accuracy = 0.0, 0.0

        for inputs_ids, attention_mask, labels in dataloader:
            outputs = model(
                input_ids=inputs_ids,
                attention_mask=attention_mask,
                labels=labels
            )
            logits = outputs.logits

            loss = criterion(logits, labels)
            logits = logits.detach().cpu().numpy()
            label_ids = labels.to("cpu").numpy()
            accuracy = calc_accuracy(logits, label_ids)

            val_loss += loss.item()
            val_accuracy += accuracy

        val_loss = val_loss / len(dataloader)
        val_accuracy = val_accuracy / len(dataloader)
        return val_loss, val_accuracy


best_loss = 10000
for epoch in range(epoch):
    train_loss = train(model, optimizer, train_loader)
    val_loss, val_accuracy = evaluation(model, valid_loader)
    print(f"Epoch {epoch+1}: Train Loss: {train_loss:.4f} Val Loss: {val_loss:.4f} Val Accuracy: {val_accuracy:.4f}")

    if val_loss < best_loss:
        best_loss = val_loss
        torch.save(model.state_dict(), "best_model.pt")
        print("Saved the model weights")
```

- 내부적으로 손실 계산하여 반환
    - Train 함수로 모델 학습 시 손실은 모델 출력값(outputs)의 loss 속성으로 가져옴.
- 모델 평가 시 교차 엔트로피 함수로 모델 평가.
    - 모델이 반환하는 손실값 아닌, 로짓(logits) 값과 레이블로 손실 계산함.
- `calc_accuracy` 함수 : 모델 정확도 계산
    - 모델의 예측 결과와 실제 레이블 비교하여 정확도 계산하고 반환.
